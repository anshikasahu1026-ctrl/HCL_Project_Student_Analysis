```python
# STUDENT PERFORMANCE AND ATTENDANCE ANALYZER
# Python + Pandas Project

import pandas as pd

# Load dataset
df = pd.read_csv('/content/student_dropout_behavior_dataset.csv')

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Convert numeric columns into numeric data
numeric_columns = [
    'quiz1_marks', 'quiz2_marks', 'quiz3_marks',
    'total_assignments', 'assignments_submitted',
    'midterm_marks', 'final_marks', 'previous_gpa',
    'total_lectures', 'lectures_attended',
    'total_lab_sessions', 'labs_attended'
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors='coerce')

# Calculate quiz average
df['quiz_average'] = df[
    ['quiz1_marks', 'quiz2_marks', 'quiz3_marks']
].mean(axis=1)

# Calculate lecture attendance percentage safely
df['lecture_attendance'] = (
    df['lectures_attended'] /
    df['total_lectures'].replace(0, pd.NA) * 100
)

# Calculate lab attendance percentage safely
df['lab_attendance'] = (
    df['labs_attended'] /
    df['total_lab_sessions'].replace(0, pd.NA) * 100
)

# Calculate assignment submission percentage safely
df['assignment_percentage'] = (
    df['assignments_submitted'] /
    df['total_assignments'].replace(0, pd.NA) * 100
)

# Main menu
while True:

    print("\n======================================")
    print(" STUDENT PERFORMANCE ANALYZER")
    print("======================================")
    print("1. Display student records")
    print("2. Display class averages")
    print("3. Display quiz analysis")
    print("4. Find top 5 students")
    print("5. Attendance analysis")
    print("6. Assignment submission analysis")
    print("7. Search student by ID")
    print("8. Performance categories")
    print("9. Exit")

    choice = input("\nEnter your choice (1-9): ").strip()

    # Option 1: Display student records
    if choice == '1':

        print("\nSTUDENT RECORDS")
        print(df.head(20).to_string(index=False))
        print("\nShowing first 20 students.")

    # Option 2: Class averages
    elif choice == '2':

        print("\nCLASS AVERAGES")

        print("Average Midterm Marks:",
              round(df['midterm_marks'].mean(), 2))

        print("Average Final Marks:",
              round(df['final_marks'].mean(), 2))

        print("Average Previous GPA:",
              round(df['previous_gpa'].mean(), 2))

    # Option 3: Quiz analysis
    elif choice == '3':

        print("\nQUIZ ANALYSIS")

        print("Quiz 1 Average:",
              round(df['quiz1_marks'].mean(), 2))

        print("Quiz 2 Average:",
              round(df['quiz2_marks'].mean(), 2))

        print("Quiz 3 Average:",
              round(df['quiz3_marks'].mean(), 2))

        print("Overall Quiz Average:",
              round(df['quiz_average'].mean(), 2))

    # Option 4: Top 5 students
    elif choice == '4':

        top_students = df.sort_values(
            by='final_marks',
            ascending=False
        ).head(5)

        print("\nTOP 5 STUDENTS BASED ON FINAL MARKS")

        print(top_students[
            ['student_id', 'name', 'final_marks']
        ].to_string(index=False))

    # Option 5: Attendance analysis
    elif choice == '5':

        print("\nATTENDANCE ANALYSIS")

        print("Average Lecture Attendance:",
              round(df['lecture_attendance'].mean(), 2), "%")

        print("Average Lab Attendance:",
              round(df['lab_attendance'].mean(), 2), "%")

        # Find students with low attendance
        low_attendance = df[
            df['lecture_attendance'] < 75
        ]

        print("\nSTUDENTS WITH LECTURE ATTENDANCE BELOW 75%")

        if not low_attendance.empty:

            print(low_attendance[
                ['student_id', 'name', 'lecture_attendance']
            ].to_string(index=False))

        else:

            print("No students have lecture attendance below 75%.")

        print("\nNumber of students below 75%:",
              len(low_attendance))

    # Option 6: Assignment analysis
    elif choice == '6':

        print("\nASSIGNMENT SUBMISSION ANALYSIS")

        # Check whether submission data is available
        if df['assignments_submitted'].notna().any():

            valid_data = df[
                df['assignments_submitted'].notna()
            ]

            total_assigned = valid_data['total_assignments'].sum()
            total_submitted = valid_data['assignments_submitted'].sum()

            if total_assigned > 0:

                submission_rate = (
                    total_submitted / total_assigned * 100
                )

                print("Total Assignments Given:",
                      total_assigned)

                print("Total Assignments Submitted:",
                      total_submitted)

                print("Overall Submission Rate:",
                      round(submission_rate, 2), "%")

            print("\nSTUDENT-WISE ASSIGNMENT SUBMISSION")

            print(valid_data[
                ['student_id', 'name',
                 'total_assignments', 'assignments_submitted',
                 'assignment_percentage']
            ].head(20).to_string(index=False))

        else:

            print("Assignment submission data is missing")
            print("in the dataset for all students.")
            print("Submission rate cannot be calculated.")

            print("\nTotal students:", len(df))
            print("Students with submission data:",
                  df['assignments_submitted'].notna().sum())
            print("Students with missing submission data:",
                  df['assignments_submitted'].isna().sum())

    # Option 7: Search student by ID
    elif choice == '7':

        sid = input("Enter Student ID: ").strip()

        student = df[
            df['student_id'].astype(str).str.strip() == sid
        ]

        if not student.empty:

            print("\nSTUDENT DETAILS")
            print(student.to_string(index=False))

        else:

            print("\nStudent not found. Please check the ID.")

    # Option 8: Performance categories
    elif choice == '8':

        def categorize(marks):

            if pd.isna(marks):
                return "Marks Not Available"

            elif marks >= 40:
                return "High Performance"

            elif marks >= 30:
                return "Average Performance"

            else:
                return "Needs Improvement"

        df['performance_category'] = df[
            'final_marks'
        ].apply(categorize)

        print("\nPERFORMANCE CATEGORIES")

        print(df[
            ['student_id', 'name',
             'final_marks', 'performance_category']
        ].head(20).to_string(index=False))

        print("\nCATEGORY COUNT")

        print(df['performance_category'].value_counts())

    # Option 9: Exit
    elif choice == '9':

        print("\nThank you for using Student Performance Analyzer!")
        break

    # Invalid input
    else:

        print("\nInvalid choice. Please enter a number from 1 to 9.")
```
